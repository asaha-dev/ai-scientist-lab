```python

import tiktoken

enc = tiktoken.get_encoding("gpt2")

text = (
    "The patient has an EGFR mutation. "
    "The mutation may affect treatment response. "
    "Molecular testing can help characterize the tumor."
)

token_ids = enc.encode(text)

context_length = 8
stride = 4

for start in range(0, len(token_ids) - context_length, stride):
    x = token_ids[start:start + context_length]
    y = token_ids[start + 1:start + context_length + 1]

    print("INPUT: ", enc.decode(x))
    print("TARGET:", enc.decode(y))
    print()
```
